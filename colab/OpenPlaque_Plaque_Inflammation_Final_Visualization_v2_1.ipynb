# OpenPlaque — Plaque + Inflammation Final Visualization v2.1

Corrected **Runtime → Run all** notebook.

This version preserves all accepted plaque and direct-PCAT endpoint results and fixes the RCA voxel-decomposition endpoint convention discovered during the v2.0 validation run.

The locked RCA research PCAT endpoint is defined by forty half-open 1-mm bins from **[10,11) through [49,50) mm**. The broader canonical shell sampler also contains voxels assigned exactly to the 50-mm endpoint station. v2.1 excludes those endpoint-cap voxels so the source-space RCA voxel reconstruction matches the locked research endpoint exactly.

The notebook:

- reproduces LAD/RCA/LCX/LM plaque best estimates;
- fixes plaque colors and legend;
- uses exact locked PCAT segment lengths;
- computes true source-space PCAT fat-voxel HU distributions;
- validates RCA/LAD/LCX voxel populations against their locked endpoints;
- produces voxel HU decomposition bars and histograms;
- retains longitudinal profiles and heatmap;
- recomputes RCA radial QC from the same locked [10,50) voxel population;
- displays all figures inline and exports all outputs to one ZIP.

**Research boundaries:** OpenPlaque plaque values are research best-estimate proxies, not Cleerly outputs. Direct PCAT is not proprietary Caristo FAI-Score. LM inflammation is not standardized. C6 remains an LCX-like structural research segment.
