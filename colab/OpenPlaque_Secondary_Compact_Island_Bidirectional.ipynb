```python
from IPython.display import display, Image
from openplaque.secondary_compact_island_bidirectional import synthetic_island_self_test
from openplaque.secondary_compact_island_bidirectional_v2 import SecondaryCompactIslandBidirectionalWorkflow
self_test=synthetic_island_self_test(); display(self_test); assert self_test['passed'], self_test
wf=SecondaryCompactIslandBidirectionalWorkflow(root='/content/drive/MyDrive/OpenPlaque', reuse={
    'inputs':REUSE_INPUTS,'seed':REUSE_SEED,'tracking':REUSE_TRACKING,'figures':REUSE_FIGURES,'report':REUSE_REPORT
})
display(wf.cache_status())
```

```python
snapshot=wf.load_inputs(); display(snapshot); display(wf.calibration)
```

```python
seed=wf.find_seed(); display(seed)
print('Top seed hypotheses:')
display(wf.seed_candidates.head(20))
```

```python
summary=wf.run_tracking(); display(summary)
print('Backward track:'); display(wf.backward)
print('Forward track:'); display(wf.forward)
print('Terminal diagnostics:'); display(wf.terminal)
```

```python
names=wf.make_figures()
for name in names:
    print(name)
    display(Image(filename=str(wf.out/name)))
```
