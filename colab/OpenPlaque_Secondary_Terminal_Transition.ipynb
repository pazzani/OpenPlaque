```python
from IPython.display import display, Image
from openplaque.secondary_terminal_transition import SecondaryTerminalTransitionWorkflow, synthetic_transition_self_test

t = synthetic_transition_self_test()
display(t)
assert t['passed'], t

wf = SecondaryTerminalTransitionWorkflow(
    root='/content/drive/MyDrive/OpenPlaque',
    reuse={'inputs': REUSE_INPUTS, 'profile': REUSE_PROFILE, 'figures': REUSE_FIGURES, 'report': REUSE_REPORT},
)
display(wf.cache_status())
```

```python
snapshot = wf.load_inputs()
display(snapshot)
display(wf.calibration)
```

```python
summary = wf.run_profile(start_arc_mm=11.5, end_arc_mm=14.5, step_mm=0.10)
display(summary)
display(wf.profile)
```

```python
names = wf.make_figures()
for name in names:
    print(name)
    display(Image(filename=str(wf.out / name)))
```
