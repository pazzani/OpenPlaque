# OpenPlaque — LCX Distal Rollback Relaunch

Freeze the accepted LAD/takeoff/trunk and 13.8-mm independent secondary branch. Roll back 1.5–4 mm from the broadened distal endpoint and relaunch several distal continuation searches from compact coronary-sized source-CCTA segments. No global rediscovery and no automatic LCX label. Research use only.

```
# Reuse controls — True + valid cache reuses; False forces recomputation/overwrite.
REUSE_SOURCE_CT = True
REUSE_FROZEN_GEOMETRY = True
REUSE_ANCHORS = True
REUSE_SEARCH = True
REUSE_FIGURES = True
REUSE_REPORT = True
```

```
%pip -q install scipy matplotlib pandas
print('Dependencies ready.')
```

## Step 4 — Load the fresh branch


```
import os, sys, subprocess, shutil
REPO='/content/OpenPlaque'
BRANCH='lcx-distal-relaunch-from-main'
if os.path.exists(REPO): shutil.rmtree(REPO)
subprocess.run(['git','clone','-q','--depth','1','--branch',BRANCH,'https://github.com/pazzani/OpenPlaque.git',REPO],check=True)
sys.path.insert(0,os.path.join(REPO,'src'))
print('Loaded',BRANCH)
```

## Step 5 — Initialize and inspect cache status


```
from openplaque.lcx_distal_relaunch import LCXDistalRelaunchWorkflow
reuse={'source_ct':REUSE_SOURCE_CT,'frozen_geometry':REUSE_FROZEN_GEOMETRY,'anchors':REUSE_ANCHORS,'search':REUSE_SEARCH,'figures':REUSE_FIGURES,'report':REUSE_REPORT}
wf=LCXDistalRelaunchWorkflow(reuse=reuse)
display(wf.cache_status())
```

## Step 6 — Load source CCTA and freeze accepted anatomy


```
wf.load_source_ct()
frozen=wf.load_frozen_geometry()
display(frozen)
display(wf.seed_qc.tail(10))
```

## Step 7 — Select compact rollback anchors


```
anchors=wf.select_anchors()
display(anchors)
```

## Step 8 — Relaunch distal continuation from each anchor


```
summary=wf.search(max_new_mm=24.0,beam_width=32)
display(summary)
display(wf.candidates.head(15) if wf.candidates is not None and len(wf.candidates) else wf.candidates)
```

## Step 9 — Figures, report, and report-back ZIP


```
wf.make_figures()
html=wf.make_report()
z=wf.package()
print('Report:',html)
print('ZIP:',z)
print('Run complete. Ask ChatGPT: Retrieve and analyze')
```
