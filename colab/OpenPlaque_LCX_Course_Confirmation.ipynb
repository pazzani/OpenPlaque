## Step 4 — Load this fresh branch


```python
import os, sys, subprocess, shutil
REPO='/content/OpenPlaque'
BRANCH='lcx-course-confirmation-from-main'
if os.path.exists(REPO): shutil.rmtree(REPO)
subprocess.run(['git','clone','-q','--depth','1','--branch',BRANCH,'https://github.com/pazzani/OpenPlaque.git',REPO],check=True)
sys.path.insert(0,os.path.join(REPO,'src'))
print('Loaded',BRANCH)
```

## Step 5 — Initialize and inspect caches


```python
from openplaque.lcx_course_confirmation import LCXCourseConfirmationWorkflow
reuse={
 'source_ct':REUSE_SOURCE_CT,
 'frozen_seed':REUSE_FROZEN_SEED,
 'course_extension':REUSE_COURSE_EXTENSION,
 'figures':REUSE_FIGURES,
 'report':REUSE_REPORT,
}
wf=LCXCourseConfirmationWorkflow(reuse=reuse)
display(wf.cache_status())
```

## Step 6 — Freeze the already accepted secondary branch and established LAD/trunk geometry


```python
wf.load_source_ct()
seed_info=wf.load_frozen_seed()
seed_info
```

## Step 7 — Extend only the accepted secondary branch in source CCTA


```python
summary=wf.extend_course(max_extra_mm=24.0,beam_width=18)
display(summary)
display(wf.candidates.head(12))
```
