# OpenPlaque — Canonical source-CCTA coronary centerline bundle

This notebook creates a clean downstream anatomy bundle after the left-coronary reconciliation experiment.

It keeps the validated RCA, selects a ~57 mm LAD only if it independently passes source-CCTA and both TotalSegmentator coronary-model gates, and selects a ~13–15 mm secondary coronary-like trajectory. The secondary trajectory is stored in full for geometry, while quantitative compact-lumen analyses are explicitly truncated at 12.2 mm.

The stale ~117 mm LAD and ~113 mm LCX-named source files are not eligible for canonical selection.
