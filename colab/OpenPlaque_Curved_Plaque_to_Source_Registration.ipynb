# OpenPlaque — curved plaque series → source-CCTA registration

This experiment tests whether the vessel-specific curved DICOM series used by the plaque nnU-Net can be mapped back into source-CCTA physical space. It first audits DICOM geometry, then fits a longitudinal mapping using source-centerline HU structure, tests 180° and 360° rotation conventions against source-generated CPR ribbons, and creates source-space plaque-support maps **only if conservative validation gates pass**.

If the curved series is a rotation stack rather than a true spatial 3-D volume, native stack `mm³` is explicitly *not* treated as anatomical plaque volume.
